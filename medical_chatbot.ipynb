```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
df= pd.read_csv('Training.csv')
```

```python
df
```

```python
le = LabelEncoder()
df["prognosis"]= le.fit_transform(df["prognosis"])
df.head()
```

```python
df
```

```python
df = df.drop(columns=['Unnamed: 133'])
display(df.head())
```

```python
precautions = pd.read_csv("symptom_precaution.csv")
df.head()
```

```python
print(precautions.info())
print('\nMissing values in precautions DataFrame:')
print(precautions.isnull().sum())
```

```python
precautions = precautions.drop(columns=['Precaution_3', 'Precaution_4'])
print(precautions.info())
print('\nMissing values in precautions DataFrame:')
print(precautions.isnull().sum())
```

```python
display(df.head())
```

```python
selected_columns = ['itching', 'shivering', 'acidity', 'stomach_pain', 'vomiting', 'fatigue', 'joint_pain', 'high_fever', 'dehydration', 'headache', 'cough', 'prognosis']
df_filtered = df[selected_columns]
display(df_filtered.head())
```

```python
X_filtered = df_filtered.drop('prognosis', axis=1)
y_filtered = df_filtered['prognosis']

model_filtered = DecisionTreeClassifier()
model_filtered.fit(X_filtered, y_filtered)

print(f"Model retrained with {len(X_filtered.columns)} selected symptoms.")
```

```python
def chatbot_():
    print("\n🩺 SMART MEDICAL ANALYSIS CHATBOT (Filtered Symptoms)")
    print("Answer symptoms with YES or NO\n")

    symptoms = {}

    for column in X_filtered.columns:
        ans = input(f"Do you have {column}? (yes/no): ").lower()
        symptoms[column] = 1 if ans == "yes" else 0

    input_data = pd.DataFrame([symptoms])

    prediction_code = model_filtered.predict(input_data)[0]
    predicted_disease = le.inverse_transform([prediction_code])[0]
    print("\n✅ Possible Disease:", predicted_disease)

    # Lookup precautions
    precaution_info = precautions[precautions["Disease"] == predicted_disease]
    if not precaution_info.empty:
        precaution_1 = precaution_info["Precaution_1"].values[0]
        precaution_2 = precaution_info["Precaution_2"].values[0]
        print("💊 Suggested Precaution 1:", precaution_1)
        print("💊 Suggested Precaution 2:", precaution_2)
    else:
        print("No specific precautions found for this disease.")

chatbot_()
```
